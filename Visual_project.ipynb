```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('/content/IPL.csv')
df
df.shape
pd.DataFrame(df)
df

```

```python
#who win the most matches
match_wins = df['match_winner'].value_counts()
# match_wins
# sns.countplot(x='match_winner',data=df)
# plt.xticks(rotation=90)
sns.barplot(x=match_wins.index,y=match_wins.values)
plt.xticks(rotation=90)
plt.show()
#
```

```python
#from datetime import d
mahina = df['venue'].value_counts()
sns.barplot(x=mahina.index,y=mahina.values,estimator=np.sum)
plt.xticks(rotation=90)
plt.title("Most match win by ")
```

```python
#Mostly TOsss decison
sns.countplot(x='toss_decision',data=df)
plt.show()
```

```python
#Match vs toss winer
cont= df[df['toss_winner'] == df['match_winner']]['match_id'].count()
Percentage = (cont/df['match_id'].count())*100
print(Percentage)
```

```python
#Most player win the match
count = df['player_of_the_match'].value_counts()
count = count.head(10)
sns.barplot(y=count.index,x=count.values,palette='mako')
plt.xticks(rotation=90)
plt.show()
```

```python
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False)
high = high.head(10)
sns.barplot(x=high.index,y=high.values)
plt.xticks(rotation=90)
plt.show()
```
